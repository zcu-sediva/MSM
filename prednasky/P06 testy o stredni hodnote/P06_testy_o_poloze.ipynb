
# Testy o poloze

Tento výukový notebook je určen pro studenty navazujícího stupně vysoké školy. Cílem je představit testy o poloze pro jednorozměrná a vícerozměrná data, jejich matematické formulace, předpoklady a praktické použití.



## Jednorozměrná data

### Jeden výběr
- **Z-test**: Používá se, pokud je známý rozptyl \( \sigma^2 \).
  - Testovací statistika: \( Z = 
rac{ar{X} - \mu_0}{\sigma / \sqrt{n}} \)
- **T-test**: Používá se, pokud \( \sigma^2 \) není známý.
  - Testovací statistika: \( T = 
rac{ar{X} - \mu_0}{s / \sqrt{n}} \)
- **Wilcoxonův test**: Neparametrický test pro medián, pokud není splněna normalita.


```python

import scipy.stats as stats
import numpy as np

# Ukázková data
data = np.array([12, 15, 14, 10, 13, 15, 14])

# T-test pro průměr
t_stat, p_value = stats.ttest_1samp(data, popmean=13)
print(f"T-statistika: {t_stat:.3f}, p-hodnota: {p_value:.3f}")
```


### Dva nezávislé výběry
- **Dvouvýběrový t-test**: Testuje rovnost průměrů dvou skupin.
  - \( T = 
rac{ar{X}_1 - ar{X}_2}{\sqrt{s_p^2 (
rac{1}{n_1} + 
rac{1}{n_2})}} \)
- **Test o shodě rozptylů**: Např. Leveneho test.


```python

# Dva výběry
group1 = np.array([10, 12, 11, 13, 12])
group2 = np.array([14, 15, 13, 16, 15])

# Dvouvýběrový t-test
t_stat, p_value = stats.ttest_ind(group1, group2)
print(f"T-statistika: {t_stat:.3f}, p-hodnota: {p_value:.3f}")
```


### Více výběrů
- **ANOVA one way**: Testuje rovnost průměrů mezi více skupinami.
- **Kruskal-Wallisův test**: Neparametrická alternativa k ANOVA.


```python

# Více skupin
group_a = [12, 14, 13]
group_b = [15, 16, 14]
group_c = [10, 11, 12]

# ANOVA
f_stat, p_value = stats.f_oneway(group_a, group_b, group_c)
print(f"F-statistika: {f_stat:.3f}, p-hodnota: {p_value:.3f}")
```
