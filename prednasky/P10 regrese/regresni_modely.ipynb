## Teorie
Lineární regresní model:
\[ Y = Xeta + arepsilon \]
Kde:
- \( Y \) je vektor závislé proměnné
- \( X \) je matice vysvětlujících proměnných
- \( eta \) je vektor parametrů
- \( arepsilon \) je náhodná složka

Odhad parametrů metodou nejmenších čtverců:
\[ \hat{eta} = (X^T X)^{-1} X^T Y \]


```
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Simulace dat
np.random.seed(42)
X = np.random.rand(100, 1) * 10
y = 3 + 2 * X.flatten() + np.random.randn(100) * 2

# Přidání konstanty (intercept)
X_const = sm.add_constant(X)

# Odhad modelu metodou OLS
model = sm.OLS(y, X_const)
results = model.fit()
print(results.summary())

# Vizualizace
plt.scatter(X, y, label='Data')
plt.plot(X, results.predict(X_const), color='red', label='Regresní přímka')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Simulace dat
np.random.seed(42)
X1 = np.random.rand(100) * 10
X2 = X1 * 0.8 + np.random.rand(100) * 2  # korelovaná proměnná
X3 = np.random.rand(100) * 10
y = 5 + 1.5*X1 + 2*X2 - 0.5*X3 + np.random.randn(100) * 3

df = pd.DataFrame({'X1': X1, 'X2': X2, 'X3': X3})
X_const = sm.add_constant(df)
model = sm.OLS(y, X_const).fit()
print(model.summary())

# Výpočet VIF
vif_values = [variance_inflation_factor(X_const.values, i) for i in range(1, X_const.shape[1])]
for col, vif in zip(df.columns, vif_values):
    print(f"{col}: VIF = {vif:.2f}")
```

```
import seaborn as sns
from scipy.stats import chi2

# Korelační matice
corr_matrix = df.corr()
print(corr_matrix)
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Korelační matice regresorů')
plt.show()

# Farrar-Glauber test
n, k = df.shape
R = corr_matrix.values
det_R = np.linalg.det(R)
T = -(n - 1 - (2*k + 5)/6) * np.log(det_R)
nu = k*(k-1)/2
p_val = chi2.sf(T, nu)
print(f"Chi-square test statistic T = {T:.3f}, df = {nu}, p = {p_val:.3g}")
```
