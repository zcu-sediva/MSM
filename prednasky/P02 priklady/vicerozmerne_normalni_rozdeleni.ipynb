
# Vícerozměrné normální rozdělení

Tento notebook ukazuje, že marginální a podmíněná rozdělení vícerozměrného normálního rozdělení jsou opět normální.

## Teorie

Mějme náhodný vektor \( \mathbf{X} = (X_1, X_2)^T \), který má vícerozměrné normální rozdělení:
\[ \mathbf{X} \sim \mathcal{N}(oldsymbol{\mu}, oldsymbol{\Sigma}) \]

Pak:
- Marginální rozdělení \( X_1 \) a \( X_2 \) jsou normální.
- Podmíněné rozdělení \( X_1 | X_2 = x_2 \) je také normální.

V následujících buňkách si to ukážeme pomocí simulací.


```python

import numpy as np
np.random.seed(42)

# Parametry rozdělení
mu = [0, 0]
Sigma = [[1, 0.8], [0.8, 1]]

# Generování vzorku
X = np.random.multivariate_normal(mu, Sigma, size=5000)
X1 = X[:, 0]
X2 = X[:, 1]
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(X1, kde=True, bins=30, color='skyblue')
plt.title("Marginální rozdělení X1")

plt.subplot(1, 2, 2)
sns.histplot(X2, kde=True, bins=30, color='salmon')
plt.title("Marginální rozdělení X2")

plt.tight_layout()
plt.show()
```

```python

# Vybereme vzorky, kde X2 je blízko 1
mask = (X2 > 0.95) & (X2 < 1.05)
X1_given_X2 = X1[mask]

plt.figure(figsize=(6, 4))
sns.histplot(X1_given_X2, kde=True, bins=30, color='green')
plt.title("Podmíněné rozdělení X1 | X2 ≈ 1")
plt.xlabel("X1")
plt.show()
```


## Shrnutí

Z výše uvedených simulací vidíme:
- Histogramy marginálních rozdělení X1 a X2 mají tvar normálního rozdělení.
- Histogram podmíněného rozdělení X1 | X2 ≈ 1 má také tvar normálního rozdělení.

To potvrzuje teoretické tvrzení, že marginální i podmíněná rozdělení vícerozměrného normálního rozdělení jsou opět normální.
